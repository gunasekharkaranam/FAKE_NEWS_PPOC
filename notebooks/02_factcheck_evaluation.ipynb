## Try your own text

```
from fakenews.pipeline import FakeNewsChecker
checker = FakeNewsChecker()
r = checker.check("The Statue of Liberty was a gift from Germany to the United States.")
print(r['verdict'], '|', r['reason'])
for c in r['claims']:
    print('\nCLAIM:', c['claim'], '->', c['verdict'])
    for e in c['evidence']:
        print('  ', e['verdict'], f"({e['method']})", e['title'], '|', e['text'][:120])
```
